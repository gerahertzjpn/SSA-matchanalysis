```python
import json

with open("../replays/01_league_league_16147200961505151604.json", "r", encoding="utf-8") as f:
    data = json.load(f)
```

```python
print(data["ticks"][0]["baron"].keys())
```

```python
exclusive = set()
for tick in data["ticks"]:
    try:
        exclusive.add(tick["baron"]["allied_lord_owner"])
    except KeyError:
        pass
print(exclusive)
```

```python
lid = {"A1", "A2", "A3", "B1", "B2", "B3"}
capturelordtick = 0
owner = ""
for tick in data["ticks"]:
    try:
        if tick["baron"]["allied_lord_owner"] in lid:
            capturelordtick = tick["tick"]
            owner = tick["baron"]["allied_lord_owner"]
    except KeyError:
        pass
print(capturelordtick)
print(owner)
```
